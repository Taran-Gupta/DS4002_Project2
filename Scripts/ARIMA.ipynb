```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.stattools import adfuller
```

```python
data_new_worth = pd.read_csv("/content/net_worth_cleaned.csv", encoding='ISO-8859-1')
data_unemployment = pd.read_csv("/content/unemployment_cleaned.csv", encoding='ISO-8859-1')
```

```python
data_new_worth.index = pd.to_datetime(data_new_worth.index)
data_unemployment.index = pd.to_datetime(data_unemployment.index)

data_new_worth, data_unemployment = data_new_worth.align(data_unemployment, join='inner', axis=0)

data_new_worth = data_new_worth.asfreq('QS')
data_unemployment = data_unemployment.asfreq('QS')
```

```python
model_fits = {}
for y_col in data_new_worth.columns:
    for x_col in data_unemployment.columns:
        y = data_new_worth[y_col].dropna()
        X = data_unemployment[[x_col]].loc[y.index]
        try:
            model = ARIMA(y, exog=X, order=(1,1,1))
            model_fit = model.fit()
            model_fits[(y_col, x_col)] = model_fit
            print(f"Fitted: {y_col} ~ {x_col}")
        except Exception as e:
            print(f"Skipped {y_col} ~ {x_col}: {e}")
```

```python
predictions = {}
for (y_col, x_col), model_fit in model_fits.items():
    last_val = data_unemployment[x_col].iloc[-1]
    future_X = pd.DataFrame({x_col: [last_val]*5})
    try:
        forecast = model_fit.forecast(steps=5, exog=future_X)
        predictions[(y_col, x_col)] = forecast
    except Exception as e:
        print(f"Forecast failed for {y_col} ~ {x_col}: {e}")
```

```python
for (y_col, x_col), pred in predictions.items():
    print(f"{y_col} predicted by {x_col}:")
    print(pred.values)
```
