## 1.0 Unemployment Data Cleaning

1.1 Import Data

```python
import pandas as pd
unemployment = pd.read_csv("../Data/unemployment by month.csv", index_col = 0)
unemployment.head()
```

1.2 Rename Columns

```python
unemployment = unemployment.rename(columns = {"observation_date": "date", "LNS14000036": "20-24", "LNS14000089": "25-34", "LNS14000091": "35-44", "LNS14000093": "45-54", "LNS14024230": "55 & Over"})

unemployment.head()
```

```python
# length of dataset
len(unemployment)
```

1.3 Aggregate Data into Quarters

```python
# Dropping all dates before 2004.
unemployment = unemployment[unemployment.index >= "2004-01-01"]


# Making it quarterly so that it matches net worth data. Setting the resample to start of quarter.
unemployment.index = pd.to_datetime(unemployment.index)
unemployment_quarterly = unemployment.resample("QS").mean()

unemployment_quarterly.head(10)
```

1.4 Observe and Check for NaNs/errors

```python
# round to 2 decimal places
unemployment_quarterly = unemployment_quarterly.round(2)
unemployment_quarterly.head(10)
```

```python
# check for NaN values
unemployment_quarterly.isna().sum()
```

1.5 Save and Export Cleaned Data

```python
unemployment_quarterly.to_csv("../Data/unemployment_cleaned.csv")
```

## 2.0 Net Worth Data Cleaning

2.1 Import Data

```python
net_worth = pd.read_csv("../Data/net worth by quarter.csv", encoding='ISO-8859-1')
net_worth.head()
```

```python
len(net_worth)
```

2.2 Rename Columns

```python
net_worth.rename(columns={'WFRBLB50107': '0% to 50%'}, inplace=True)
net_worth.rename(columns={'WFRBLN40080': '50% to 90%'}, inplace=True)
net_worth.rename(columns={'WFRBLN09053': '90% to 99%'}, inplace=True)
net_worth.rename(columns={'WFRBL99T999219': '99% to 99.9%'}, inplace=True)
net_worth.rename(columns={'WFRBLTP1246': '99.9% to 100%'}, inplace=True)
net_worth.rename(columns={'WFRBLT01026': '99% to 100%'}, inplace=True)

# drop overlapping 99% to 100% column
net_worth.drop(columns=['99% to 100%'], inplace=True)
```

2.3 Rename Dates to Quarters (or match 1.3)

```python
net_worth['observation_date'] = pd.to_datetime(net_worth['observation_date'])
net_worth = net_worth[(net_worth['observation_date'].dt.year >= 2004)]
```

2.4 Observe and Check for NaNs/errors

```python
net_worth.head()
```

```python
# check for errors
net_worth.isna().sum()
```

2.5 Save and Export Cleaned Data

```python
net_worth.to_csv("../Data/net_worth_cleaned.csv", index=False)
```
